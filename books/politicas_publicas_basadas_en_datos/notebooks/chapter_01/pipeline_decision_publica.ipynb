# Capitulo 01 - Pipeline de decision publica

Este notebook crea una matriz didactica para conectar problemas publicos, indicadores, usos de evidencia y advertencias. Los datos son simulados y no representan resultados reales.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd


def find_book_root() -> Path:
    cwd = Path.cwd().resolve()
    for base in [cwd, *cwd.parents]:
        nested = base / "books" / "politicas_publicas_basadas_en_datos"
        if nested.exists():
            return nested
        if base.name == "politicas_publicas_basadas_en_datos" and (base / "PROJECT_BRIEF.md").exists():
            return base
    raise FileNotFoundError("No se encontro la raiz del libro.")


BOOK_ROOT = find_book_root()
TABLES_DIR = BOOK_ROOT / "outputs" / "tables"
FIGURES_DIR = BOOK_ROOT / "outputs" / "figures"
REPORTS_DIR = BOOK_ROOT / "outputs" / "reports"

for path in [TABLES_DIR, FIGURES_DIR, REPORTS_DIR]:
    path.mkdir(parents=True, exist_ok=True)

BOOK_ROOT
```

```python
matriz = pd.DataFrame(
    [
        {
            "problema_publico": "Solicitudes no atendidas de cuidado infantil",
            "poblacion_objetivo": "Hogares con ninos pequenos en zonas urbanas",
            "indicador": "Solicitudes no atendidas por cada 1.000 ninos",
            "uso_evidencia": "Diagnostico",
            "fuente_potencial": "Registro administrativo de servicios de cuidado",
            "decision_posible": "Priorizar territorios con mayor brecha de acceso",
            "advertencia": "Revisar cobertura del registro y hogares fuera del sistema.",
        },
        {
            "problema_publico": "Abandono de tramites sociales",
            "poblacion_objetivo": "Solicitantes de transferencias sociales",
            "indicador": "Probabilidad estimada de no completar el tramite",
            "uso_evidencia": "Prediccion",
            "fuente_potencial": "Registro administrativo de postulaciones",
            "decision_posible": "Activar acompanamiento antes del vencimiento",
            "advertencia": "Auditar sesgos y no usar el modelo para negar derechos.",
        },
        {
            "problema_publico": "Baja ejecucion de cupos de capacitacion",
            "poblacion_objetivo": "Participantes inscritos en cursos publicos",
            "indicador": "Porcentaje mensual de cupos efectivamente usados",
            "uso_evidencia": "Monitoreo",
            "fuente_potencial": "Sistema de gestion del programa",
            "decision_posible": "Ajustar convocatorias y seguimiento operativo",
            "advertencia": "No confundir avance operativo con impacto laboral.",
        },
        {
            "problema_publico": "Efectividad de una intervencion educativa",
            "poblacion_objetivo": "Estudiantes elegibles para tutorias",
            "indicador": "Diferencia de resultados atribuible al programa",
            "uso_evidencia": "Evaluacion causal",
            "fuente_potencial": "Diseno experimental o cuasi-experimental documentado",
            "decision_posible": "Escalar, ajustar o cerrar la intervencion",
            "advertencia": "Requiere estrategia de identificacion y supuestos verificables.",
        },
    ]
)

matriz
```

```python
tabla_salida = TABLES_DIR / "chapter_01_matriz_decision_publica.csv"
matriz.to_csv(tabla_salida, index=False, encoding="utf-8")

conteo_usos = matriz["uso_evidencia"].value_counts().sort_index()

fig, ax = plt.subplots(figsize=(8, 4.5))
conteo_usos.plot(kind="bar", ax=ax, color=["#2f6f73", "#5e8c61", "#c28f2c", "#7d5a8c"])
ax.set_title("Usos de evidencia en la matriz didactica")
ax.set_xlabel("Tipo de uso")
ax.set_ylabel("Numero de ejemplos")
ax.tick_params(axis="x", rotation=25)
ax.grid(axis="y", alpha=0.25)
fig.tight_layout()

figura_salida = FIGURES_DIR / "chapter_01_usos_evidencia.png"
fig.savefig(figura_salida, dpi=180)

reporte_salida = REPORTS_DIR / "chapter_01_nota_metodologica.md"
reporte_salida.write_text(
    "# Nota metodologica - Capitulo 01\n\n"
    "Los datos de este notebook son simulados y pedagogicos. "
    "No describen una institucion, programa o territorio real. "
    "La matriz sirve para practicar trazabilidad entre problema, indicador, "
    "uso de evidencia, decision y advertencia.\n",
    encoding="utf-8",
)

tabla_salida, figura_salida, reporte_salida
```

## Interpretacion

La matriz separa diagnostico, monitoreo, prediccion y evaluacion causal. Esta separacion evita afirmar impacto cuando solo hay descripcion o seguimiento operativo.
