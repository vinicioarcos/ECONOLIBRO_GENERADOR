# Capitulo 2 - Verificacion de entorno

Este notebook verifica que el proyecto tenga una estructura minima reproducible. No descarga datos y no modifica `data/raw`.

## 1. Detectar la raiz del proyecto

Buscamos `PROJECT_BRIEF.md` como ancla del repositorio. Esto permite ejecutar el notebook desde distintas carpetas sin usar rutas absolutas.

```python
from pathlib import Path
import csv
import platform
import sys


def find_project_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "PROJECT_BRIEF.md").exists():
            return candidate
    raise FileNotFoundError("No se encontro PROJECT_BRIEF.md en los directorios padres.")


PROJECT_ROOT = find_project_root(Path.cwd().resolve())
DATA_RAW = PROJECT_ROOT / "data" / "raw"
OUTPUT_REPORTS = PROJECT_ROOT / "outputs" / "reports"
OUTPUT_TABLES = PROJECT_ROOT / "outputs" / "tables"

for folder in [OUTPUT_REPORTS, OUTPUT_TABLES]:
    folder.mkdir(parents=True, exist_ok=True)

print(f"Raiz del proyecto: {PROJECT_ROOT}")
print(f"Python: {sys.version.split()[0]}")
print(f"Plataforma: {platform.platform()}")
```

## 2. Checklist de archivos y carpetas

Revisamos componentes minimos del proyecto. Esta tabla ayuda a diagnosticar problemas antes de trabajar con datos reales.

```python
required_paths = [
    "PROJECT_BRIEF.md",
    "README.md",
    "requirements.txt",
    "environment.yml",
    "chapters/02_instalacion_entorno.md",
    "notebooks/chapter_02/verificacion_entorno.ipynb",
    "data/raw",
    "data/processed",
    "outputs/reports",
    "outputs/tables",
]

checklist = []
for relative_path in required_paths:
    path = PROJECT_ROOT / relative_path
    checklist.append(
        {
            "ruta": relative_path,
            "existe": path.exists(),
            "tipo": "carpeta" if path.is_dir() else "archivo",
        }
    )

for item in checklist:
    estado = "OK" if item["existe"] else "FALTA"
    print(f"{estado}: {item['ruta']}")
```

## 3. Guardar salidas reproducibles

Exportamos una tabla CSV y un reporte TXT. Ambos pueden regenerarse ejecutando el notebook desde cero.

```python
checklist_path = OUTPUT_TABLES / "chapter_02_checklist_entorno.csv"
report_path = OUTPUT_REPORTS / "chapter_02_entorno_python.txt"

with checklist_path.open("w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=["ruta", "existe", "tipo"])
    writer.writeheader()
    writer.writerows(checklist)

raw_files = [p.name for p in DATA_RAW.iterdir() if p.name != ".gitkeep"] if DATA_RAW.exists() else []
report_lines = [
    "Reporte de entorno - Capitulo 2",
    f"Raiz del proyecto: {PROJECT_ROOT.name}",
    f"Python: {sys.version.split()[0]}",
    f"Ejecutable: {sys.executable}",
    f"Plataforma: {platform.platform()}",
    f"Archivos no .gitkeep en data/raw: {len(raw_files)}",
    "Regla: no modificar data/raw; escribir derivados en data/processed u outputs.",
]

report_path.write_text("\n".join(report_lines) + "\n", encoding="utf-8")

print(f"Checklist guardado en: {checklist_path.relative_to(PROJECT_ROOT)}")
print(f"Reporte guardado en: {report_path.relative_to(PROJECT_ROOT)}")
```

## 4. Cierre

Si todas las rutas aparecen como `OK`, el proyecto esta listo para ejecutar notebooks iniciales. Si algo falta, primero corrija estructura o ubicacion de archivos antes de analizar datos economicos.
