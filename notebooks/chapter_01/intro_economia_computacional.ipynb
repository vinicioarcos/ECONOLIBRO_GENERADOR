## 1. Pregunta economica

Pregunta guia del ejercicio:

> Que fuente de datos permitiria estudiar una pregunta economica sobre empleo, ingresos, precios, educacion o pobreza?

En este notebook no se presentan resultados empiricos. Solo documentamos una fuente candidata para preparar el trabajo de capitulos posteriores.

```python
from pathlib import Path
import csv


def find_project_root(start: Path) -> Path:
    """Find the repository root from the notebook execution directory."""
    for candidate in [start, *start.parents]:
        if (candidate / "PROJECT_BRIEF.md").exists():
            return candidate
    raise FileNotFoundError("No se encontro PROJECT_BRIEF.md en los directorios padres.")


PROJECT_ROOT = find_project_root(Path.cwd().resolve())
DATA_RAW = PROJECT_ROOT / "data" / "raw"
DATA_PROCESSED = PROJECT_ROOT / "data" / "processed"
OUTPUT_TABLES = PROJECT_ROOT / "outputs" / "tables"

for folder in [DATA_RAW, DATA_PROCESSED, OUTPUT_TABLES]:
    folder.mkdir(parents=True, exist_ok=True)

print(f"Raiz del proyecto: {PROJECT_ROOT.name}")
print(f"data/raw existe: {DATA_RAW.exists()}")
print(f"outputs/tables existe: {OUTPUT_TABLES.exists()}")
```

## 2. Ficha de fuente candidata

La siguiente ficha es un ejemplo editable. Antes de descargar cualquier archivo, conviene registrar institucion, unidad de observacion, periodo y limites de interpretacion.

```python
ficha_fuente = [
    {
        "pregunta_economica": "Como describir cambios en empleo o ingresos usando una fuente oficial?",
        "fuente_candidata": "Encuesta laboral u otra fuente oficial por definir",
        "institucion": "Pendiente de verificar por el estudiante",
        "unidad_observacion": "Persona, hogar, empresa, territorio o periodo",
        "periodo": "Pendiente de documentar",
        "variables_necesarias": "Variable laboral, ingreso, edad, educacion, region, periodo",
        "producto_esperado": "Tabla descriptiva o figura con fuente e interpretacion",
        "advertencia": "No afirmar causalidad sin estrategia de identificacion",
    }
]

for clave, valor in ficha_fuente[0].items():
    print(f"{clave}: {valor}")
```

## 3. Exportacion reproducible

Guardamos la ficha como CSV en `outputs/tables`. Esta salida se puede regenerar al ejecutar el notebook desde cero.

```python
output_path = OUTPUT_TABLES / "chapter_01_ficha_fuente_datos.csv"

with output_path.open("w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=list(ficha_fuente[0].keys()))
    writer.writeheader()
    writer.writerows(ficha_fuente)

print(f"Ficha guardada en: {output_path.relative_to(PROJECT_ROOT)}")
```

## 4. Interpretacion

La ficha no contiene resultados empiricos. Su valor es metodologico: ayuda a decidir si una fuente puede responder una pregunta economica y que limites deben declararse antes de analizar datos.
